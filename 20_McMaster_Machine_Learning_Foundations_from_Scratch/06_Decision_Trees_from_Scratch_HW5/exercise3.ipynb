```python
from tensorflow.keras.datasets import mnist
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

(train_images, train_labels), _ = mnist.load_data()
train_images = train_images.reshape((60000, 28 * 28))
train_images = train_images.astype('float32') / 255

alpha = 1.
model = keras.Sequential([
  layers.Dense(512, activation='relu'),
  layers.Dense(10, activation='softmax')
])
model.compile(optimizer=keras.optimizers.RMSprop(alpha),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
history=model.fit(train_images, train_labels,
          epochs=10,
          batch_size=128,
          validation_split=0.2)


accuracy = history.history['accuracy']
val_accuracy = history.history['val_accuracy']

plt.figure()
plt.plot(range(len(accuracy)), accuracy)
plt.plot(range(len(val_accuracy)), val_accuracy)
plt.title("Accuracy vs Epochs alpha=1")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend(["Train", "Validation"])
plt.show()

loss = history.history['loss']
val_loss = history.history['val_loss']

plt.figure()
plt.plot(range(len(loss)), loss)
plt.plot(range(len(val_loss)), val_loss)
plt.title("Loss vs Epochs alpha=1")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.show()
```

```python
alpha = 0.001
model = keras.Sequential([
  layers.Dense(512, activation='relu'),
  layers.Dense(10, activation='softmax')
])
model.compile(optimizer=keras.optimizers.RMSprop(alpha),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
history=model.fit(train_images, train_labels,
          epochs=10,
          batch_size=128,
          validation_split=0.2)

accuracy = history.history['accuracy']
val_accuracy = history.history['val_accuracy']

plt.figure()
plt.plot(range(len(accuracy)), accuracy)
plt.plot(range(len(val_accuracy)), val_accuracy)
plt.title("Accuracy vs Epochs alpha = 0.001")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend(["Train", "Validation"])
plt.show()

loss = history.history['loss']
val_loss = history.history['val_loss']

plt.figure()
plt.plot(range(len(loss)), loss)
plt.plot(range(len(val_loss)), val_loss)
plt.title("Loss vs Epochs alpha = 0.001")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.show()
```

As you decrease the value of alpha (learning rate), the training becomes more stable but slower, the accuracy starts incresing and the loss makes sense.

For alpha = 1.0,

Train accuracy decreases because the weights jump too much and the model cannot learn consistent patterns.
Validation accuracy oscillates because predictions change randomly at each update.
Train and validation loss go to 0 due to numerical saturation of the softmax, not because the model performs well.

### Conclusion

the learning rate is too high, causing unstable updates, numerical issues in the loss, and poor accuracy. The model does not converge and the reported loss values are not reliable.

```python
model = keras.Sequential([layers.Dense(10, activation='softmax')])
model.compile(optimizer='rmsprop',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
history_small_model = model.fit(
    train_images, train_labels,
    epochs=50,
    batch_size=128,
    validation_split=0.2)
```

```python
loss = history_small_model.history['loss']
val_loss = history_small_model.history['val_loss']

plt.figure()
plt.plot(range(len(loss)), loss)
plt.plot(range(len(val_loss)), val_loss)
plt.title("Loss vs Epochs Small model")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.show()
```

The validation loss matches the training loss until around epoch 15 where the train loss decrease slightly more than the validation loss, but there is no clear sign of overfitting.

```python
model = keras.Sequential([
  layers.Dense(96, activation='relu'),
  layers.Dense(96, activation='relu'),
  layers.Dense(10, activation='softmax'),
])
model.compile(optimizer='rmsprop',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
history_large_model = model.fit(
    train_images, train_labels,
    epochs=50,
    batch_size=128,
    validation_split=0.2)
```

```python
loss = history_large_model.history['loss']
val_loss = history_large_model.history['val_loss']

plt.figure()
plt.plot(range(len(loss)), loss)
plt.plot(range(len(val_loss)), val_loss)
plt.title("Loss vs Epochs Small model")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.show()
```

The validation loss hits its minimum at around epoch 10 where the model start overfitting the training data.

From epoch 10, the training loss continue to decrease while the validation loss increases which is a clear sign of model overfitting.
