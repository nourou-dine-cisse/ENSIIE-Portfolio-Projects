```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))
    
def init_weights(n_inputs, n_hidden, n_output):
    W0 = np.random.randn(n_hidden, n_inputs)
    W1 = np.random.randn(n_hidden, n_hidden)
    W2 = np.random.randn(n_output, n_hidden)
    return W0, W1, W2
    
def feedforward(x, W0, W1, W2):
    z0 = np.dot(W0, x)
    a0 = sigmoid(z0)
    z1 = np.dot(W1, a0)
    a1 = sigmoid(z1)
    z2 = np.dot(W2, a1)
    a2 = sigmoid(z2)
    return z0, a0, z1, a1, z2, a2
    
def predict(x, W0, W1, W2):
    _,_, _, _, _, a2 = feedforward(x, W0, W1, W2)
    return a2
    
def backprop(X_train, Y_train, W0, W1, W2, n_batch, learning_rate):
    m = len(X_train)
    grad_W0 = 0.0
    grad_W1 = 0.0
    grad_W2 = 0.0
    count = 0
    for x, y in zip(X_train, Y_train):
        #Forward pass
        z0, a0, z1, a1, z2, a2 = feedforward(x, W0, W1, W2)
        #Back Propagation
        delta2 = 2 * (a2 - y) * sigmoid_derivative(z2)
        delta1 = (W2.T @ delta2) * sigmoid_derivative(z1)
        delta0 = (W1.T @ delta1) * sigmoid_derivative(z0)

        grad_W2 += np.outer(delta2, a1)/n_batch
        grad_W1 += np.outer(delta1, a0)/n_batch
        grad_W0 += np.outer(delta0, x)/n_batch
        count += 1

        if (count % n_batch == 0):
            W0 -= learning_rate * grad_W0
            W1 -= learning_rate * grad_W1
            W2 -= learning_rate * grad_W2
            grad_W0 = 0.0
            grad_W1 = 0.0
            grad_W2 = 0.0
            count = 0
            
    return W0, W1, W2

def loss(Y_pred, Y):
    return np.mean((Y_pred - Y) ** 2)

def train(X_train, Y_train, n_inputs, n_hidden, n_output,n_batch, n_epochs,learning_rate):
    W0, W1, W2 = init_weights(n_inputs, n_hidden, n_output)
    for epoch in range(n_epochs):
        W0, W1, W2 = backprop(X_train, Y_train, W0, W1, W2, n_batch, learning_rate)
        Y_pred = predict(X_train.T, W0, W1, W2)
        train_loss = loss(Y_pred.T, Y_train)
        print(f'epoch {epoch}, train loss {train_loss}')
    return W0, W1, W2, train_loss
```

```python
# Initialize network
n_batch = 200
n_samples = 20000
n_inputs = 10
n_hidden = 5
n_output = 3
W0, W1, W2 = init_weights(n_inputs, n_hidden, n_output)
```

```python
# Generate toy data
X_train = np.random.randn(n_samples,n_inputs)
Y_train = np.random.randn(n_samples,n_output)
```

```python
# Train the network
W0, W1, W2, last_loss = train(X_train, Y_train, n_inputs, n_hidden, n_output,n_batch,n_epochs=100, learning_rate=0.02)
```

```python
# Make a prediction
x_example = np.random.rand(n_inputs)
prediction = predict(x_example, W0, W1, W2)

print("Prediction:", prediction)
```

```python
model = tf.keras.models.Sequential([
  tf.keras.Input(shape=[n_inputs,]),
  tf.keras.layers.Dense(n_hidden, activation='sigmoid'),
  tf.keras.layers.Dense(n_hidden, activation='sigmoid'),
  tf.keras.layers.Dense(n_output,activation='sigmoid')
])
model.compile(
    optimizer=tf.keras.optimizers.SGD(0.02),
    loss=tf.keras.losses.MSE,
    metrics=[tf.keras.metrics.Accuracy()],
)

model.fit(
    X_train, Y_train,
    epochs=100
)
```
