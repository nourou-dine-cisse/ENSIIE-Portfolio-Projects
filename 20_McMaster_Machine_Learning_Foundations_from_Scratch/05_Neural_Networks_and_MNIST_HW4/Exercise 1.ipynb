# Exercise 1



My final project will develop a sentiment analysis pipeline for Wolof, a low-resource African language. Using **SENCORPUS**, a parallel French-Wolof corpus, I will generate weak sentiment labels by applying a pre-trained French classifier and projecting them to Wolof. Text will be represented with **character-level TF-IDF n-grams**, and **PCA** and an **autoencoder** will be used for dimensionality reduction and latent space learning. I will train **classical classifiers** (Logistic Regression, SVM, Random Forest) and a **neural network** on these representations, and also explore **unsupervised clustering** to see if sentiment structure emerges naturally. As an extension, I will build a **Wolof-to-French character-level translation model** to evaluate whether translation quality aligns with sentiment classification. This project integrates **representation learning, supervised and unsupervised ML, and sequence modelling** in a single coherent pipeline.
