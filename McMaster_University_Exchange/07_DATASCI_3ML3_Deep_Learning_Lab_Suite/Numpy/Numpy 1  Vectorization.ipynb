# What's the point ? 

```python
from math import *
def myfunc(x):
    return sin(sqrt(x))

myfunc(2.5)
```

```python
x=1.7
myfunc(x)
```

```python
a = [1.0,3.0,12.0,1.5]
print(a)
```

```python
myfunc(a)
```

Note that the array in straight python is actually represented as an array.
We want a function that can treat the entire array in one go !

```python
import numpy as np
b=np.sin(np.sqrt(a))
print(b)
```

```python
from math import *
def mynpfunc(x):
    return np.sin(np.sqrt(x))

b=mynpfunc(a)
print(b)
```

### The function has become vecorized !
### However, we need to watch out not to break the vectorization

```python
import numpy as np
np.random.seed(0)

def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output
        
values = np.random.randint(1, 10, size=5)
compute_reciprocals(values)
```

```python
big_array = np.random.randint(1, 100, size=1000000)
%timeit compute_reciprocals(big_array)
```

### Instead we should use universal functions

```python
%timeit (1.0 / big_array)
```

### Note the factor of approx 1,000 in speed 

# Lesson 1: Never loop over elements in a list in python. Always use numpy vectorization.
