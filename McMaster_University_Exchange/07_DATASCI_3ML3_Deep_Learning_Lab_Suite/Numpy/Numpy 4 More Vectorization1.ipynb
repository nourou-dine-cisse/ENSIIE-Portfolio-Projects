```python
from math import *
import numpy as np
def myfnc(x):
    return np.sum(x*x)
```

```python
a = np.arange(72.)
a.shape=(4,6,3)
print(a)
print(myfnc(a))
```

We get the sum of **all** the elements squared. We can look at this as doing the $\sum_{ijk}x_{ijk}^2$. Let's do something a little more advanced.

```python
def newfnc(x,ax):
    return np.sum(x*x,axis=ax)
```

```python
b=newfnc(a,0)
b.shape
```

With three indices the axis we specify is the one to be summed over, so that we get **b**=$\sum_{axis=i}x_{ijk}^2$, a two dimensional array with shape $(j,k)$. If we look at a two dimensional array, we can get the sum of each row by changing the axis.

```python
a = np.arange(24.)
a.shape=(4,6)
print(newfnc(a,1))
```

I want to normalize each row of the matrix **a**. How can I do that ?

```python
a /= np.sum(a*a,axis=1)
```

This doesn't work because the shape of what np.sum returned is (4,). Let's change that.

```python
a /= np.sqrt(np.sum(a*a,axis=1,keepdims=True))
print(a)
```

```python
print(np.sum(a*a,axis=1))
```

If you didn't know much about vectorization you might have written this as follows, which is completely equivalent but much slower. Try out the difference in the evaluation time for the two cells below.

```python
a = np.arange(24000000.)
a.shape=(4000,6000)
for i in range(4000):
    nrm = 0.0
    for j in range(6000):
        nrm += a[i,j]*a[i,j]
    for j in range(6000):
        a[i,j] /= sqrt(nrm)                
```
