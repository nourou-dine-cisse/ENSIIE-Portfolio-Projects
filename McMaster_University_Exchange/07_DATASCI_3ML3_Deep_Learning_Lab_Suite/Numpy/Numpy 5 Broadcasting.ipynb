# The Broadcasting Rule
### In order to broadcast, the size of the trailing axes for both arrays in an operation must either be the same size or one of them must be one. So, in the example above, if the vector b instead had been a (3,2) matrix it would not have worked.

```python
b=np.random.randn(3,2)
print(b)
c=a-b
print(a)
print(c)
```

However, if I match the indeces for higher dimensional data, it does work

```python
d=np.random.randn(3,4,4)
e=np.random.randn(3,1,4)
f=d-e
```
