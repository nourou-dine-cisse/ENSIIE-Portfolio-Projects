Let's declare a numpy array with integers from 0 to 11 and look at its shape.

```python
import numpy as np
a = np.arange(12)
a.shape
```

As we have discussed we get a **COLUMN** vector with a single index. For consistency I prefer to explicitly use 2 indeces so let's reshape it to (12,1). It is still a **COLUMN** vector.

```python
a.shape=(12,1)
```

```python
a.shape
```

We can form the transpose very easily to get an object that now has shape (1,12). If you went through the notebook on numpy storage you will know that nothing really changes when you take the transpose it is just the **view** that is modified. 

```python
b=a.T
b.shape
```

I can now form the dot product (the scalar product) between the two vectors:

```python
c=np.dot(b,a)
print(c)
print(c.shape)
type(c)
```

BTW, note that when you print out a numpy object you it prints a number of nested square brackets that corresponds to the number of indeces. So, when the system prints [[506]] we know that this an object with **two** indices even though it is just a single number.

So, what is returned is technically a numpy array with two indeces. However, I can still use it as a single number.

```python
from math import *
sqrt(c)
```

Let's think about what happened. The row vector b had shape (1,12) and the column vector a had shape (12,1) the product has shape (1,12)(12,1)=(1,1) so the two indeces of the same length of 12 has been summed over and disappeared. What happens if we multiply the other way by forming a dot b ? We would expect a **MATRIX** since (12,1)(1,12) = (12,12). Let's try:

```python
d=np.dot(a,b)
print(d.shape)
type(d)
```

Let's do that again but without explicitly reshaping to (12,1)

```python
d=np.arange(12)
print(d.shape)
e=d.T
print(e.shape)
f=np.dot(e,d)
print(f)
print(f.shape)
type(f)
```

So, in this case we get a result back which is simply an integer with no indeces. 

Let's now have a look at how matrices and vectors interact. The rule of thumb is that if we write out a product of numpy objects with the following shapes (i,j)(k,l)(m,n) then this should only work if j=k and l=m and the result should be of shape (i,n). Let's try with objects of shape (4,16)(16,2)(2,64) = (4,64)

```python
a=np.arange(64)
a.shape=(4,16)
b=np.arange(32)
b.shape=(16,2)
c=np.arange(128)
c.shape=(2,64)
d=np.dot(np.dot(a,b),c)
print(d.shape)
```

This seems to work, but had the array b been of shape (7,17) it should not work

```python
a=np.arange(64)
a.shape=(4,16)
b=np.arange(119)
b.shape=(7,17)
c=np.arange(128)
c.shape=(2,64)
d=np.dot(np.dot(a,b),c)
print(d.shape)
```

Indeed we get an error message. 

Let's now look at some simple manipulations with matrices and vectors. We start we a matrix applied to a column vector

```python
a=np.arange(64)
a.shape=(8,8)
b=np.arange(8)
b.shape=(8,1)
c=np.dot(a,b)
print(c.shape)
```

In this case we again get a column vector back. In this example if we instead form the product np.dot(b,a) we should get an error since (8,1)(8,8) do not match. Let's try:

```python
c=np.dot(b,a)
```

In this case we need to take the transpose of b to form a **ROW** vector

```python
c=np.dot(b.T,a)
print(c.shape)
```

Now we can continue and look at the **ROW** vector b.T multiplied with the matrix **a** which is multiplied with the **COLUMN** vector b. We get an object of form (1,8)(8,8)(8,1)=(1,1) - a single number.

```python
d=np.dot(np.dot(b.T,a),b)
print(d)
print(d.shape)
```

Finally, let us just note that the transpose of course extends to matrices:

```python
a=np.arange(12)
a.shape=(3,4)
print(a)
```

```python
print(a.T)
print(a.T.shape)
```

If you ever get to work with vectors and matrices and more generally tensors in physics then the above feels rather natural. However, since we're looking at a numerical implementation there are some idosynchratic details to be aware of. Notably, a single umpy array is by definition a **COLUMN** vector and when we form dot products we have to explicitly transpose. Mathematically it would have made more sense to simply **define** np.dot(a,b) to be the dot product of a.T and b. That's all details though.
