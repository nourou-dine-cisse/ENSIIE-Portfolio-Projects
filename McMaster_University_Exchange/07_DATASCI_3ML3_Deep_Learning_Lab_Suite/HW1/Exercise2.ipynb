```python
# numpy.random.randn
# importing numpy
import numpy as np

# setting the parameters
num_samples = 100
D = 10

# generating random directions
directions = np.random.randn(num_samples, D)

# testing if the rows correspond to unit vectors
print(f"Norm of the first vector: {np.linalg.norm(directions[0])}")
print(f"First direction vector: {directions[0]}")
```

2.1.1-
The element of "directions" are random numbers generated from the normal distribution.

2.1.2-
Based on the norm of first vector, the rows generated are not unit vectors.

```python
product = directions * directions
print(f"First row of the product matrix: {product[0]}")
```

the operation direction * direction results in the element-wise product of each row with itself

```python
print(f"Total sum of squares : {np.sum(product)} \n") # results in total sum of squares of all elements
print(f"sum of squares for each column: {np.sum(product,axis=0)} \n") # results in sum of squares for each column
psum = np.sum(product,axis=1) # results in sum of squares for each row
print(f"sum of squares for each row: {psum}")
```

2.3-

```python
# computing the norms and normalizing the directions
norms = np.sqrt(psum)[:,np.newaxis] # norms = np.sqrt(psum) does not work due to broadcasting issues
directions = directions/norms
```

```python
# Checking if the rows correspond to unit vectors
print(f"Norms vectors after normalization: {np.linalg.norm(directions, axis=1)}")
```
