```python
import numpy as np
from numpy import linalg as LA
import matplotlib.pyplot as plt
import torch
```

```python
def gd_with_ad(f, x0, alpha=1e-3, niters=int(1e6)):
    xk = torch.tensor(x0, requires_grad=True, dtype=torch.double)
    
    for _ in range(niters):
        value = f(xk)
        value.backward()

        with torch.no_grad():  
            xk -= alpha * xk.grad

        xk.grad.zero_()

    return xk.numpy(force=True), f(xk).item()
```

```python
def f(x):
    return x**4

print(gd_with_ad(f, 2, niters=int(1e4)))
```

```python
print(gd_with_ad(f, -2, niters=10000))
```
