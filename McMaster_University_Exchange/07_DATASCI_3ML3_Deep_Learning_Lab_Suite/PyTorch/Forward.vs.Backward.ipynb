```python
import numpy as np
from numpy import linalg as LA
import torch
```

Forward evaluation

```python
x = torch.tensor([1.,0.,-1.], requires_grad=True)
y = torch.tensor([0.,1.])
W0 = torch.tensor([[0.,1.,-1.],[2.,0.,1.]])
W1 = torch.tensor([[-1.,0.],[2.,-1.]])

z0 = x
z1 = W0 @ z0
z2 = W1 @ z1
f = 0.5 * (torch.linalg.vector_norm(y-z2) ** 2)

print(z0)
```

```python
print(z1)
```

```python
print(z2)
```

Forward derivative:

```python
with torch.no_grad():
    F0 = W0
    F1 = W1 @ F0
    grad_f = F1.T@(z2 - y)
    
print(F0)
```

```python
print(F1)
```

```python
print(grad_f)
```

Let's do a check

```python
f.backward()
print(x.grad)
```

It Works !

Now in the Revers/Backward direction

```python
with torch.no_grad():
    G2 = (z2 - y).unsqueeze(0)
    G1 = G2 @ W1
    grad_f = G1 @ W0
    
print(G2)
```

```python
print(G1)
```

```python
print(grad_f)
```

Same Result !
